## 3. Apply to Real Data (1000 Genomes Subset) (25 points)

Your goal in this section is to try to create a figure similar to Figure 1 of the Diaz-Papkovich *et al.* (2019) paper, using a subset of genotype data from the 1000 Genomes Project.

To do so, complete the following steps:

a. Load the genotype dataset accompanying this lab, where each row corresponds to an individual, and each column to a genetic variant. Load also the provided population labels for each individual.

b. **Standardize** the features (i.e., transform the genotype values to have zero mean and unit variance).

c. Apply **PCA**, **t-SNE**, and **UMAP** to reduce the data to 2 dimensions. You may use the default values for the rest of hyperparameters in these techniques.

d. Create **scatter plots** of the resulting projections, coloring each point according to its **population group**.

e. Reflect on the following questions:
- Explain what you observe in each case. Additionally, which method appears to provide the best visual separation of the population into distinct groups? Elaborate on why this might be the case, based on what you know about each method.
- What changes if you first reduce the data to the **top 25 PCA components**, and then apply t-SNE or UMAP?
- In your opinion, what are the trade-offs between **interpretability** (e.g., preservation of local and global structure) and **performance** (e.g., visual separation of known populations and computational time) when using these dimensionality reduction methods?

Please elaborate on your answers and use the plots to support your discussion.

```python
from google.colab import files
uploaded = files.upload()

import time
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import umap

# a. Load data
X = np.load('genome_data_lab3.npy').astype(float)
with open("population_labels_lab3.txt", "r", encoding="utf-8") as file:
    pop = np.array([line.strip() for line in file])

print("Genotype matrix:", X.shape)
print("Labels:", pop.shape, "| Populations:", len(np.unique(pop)))

X[X == -1] = np.nan
X = np.where(np.isnan(X), np.nanmean(X, axis=0), X)

# b. Standardize
X_std = StandardScaler().fit_transform(X)

# c. PCA, t-SNE, UMAP
def run(name, model, data):
    t0 = time.time()
    emb = model.fit_transform(data)
    elapsed = time.time() - t0
    print(f"{name:<22}{elapsed:6.1f} s")
    return emb, elapsed

results = {}
results["PCA"]   = run("PCA",   PCA(n_components=2, random_state=42), X_std)
results["t-SNE"] = run("t-SNE", TSNE(n_components=2, random_state=42), X_std)
results["UMAP"]  = run("UMAP",  umap.UMAP(n_components=2, random_state=42), X_std)

t0 = time.time()
X_pc25 = PCA(n_components=25, random_state=42).fit_transform(X_std)
pca25_time = time.time() - t0
emb, t = run("t-SNE on 25 PCs", TSNE(n_components=2, random_state=42), X_pc25)
results["t-SNE on top 25 PCs"] = (emb, t + pca25_time)
emb, t = run("UMAP on 25 PCs", umap.UMAP(n_components=2, random_state=42), X_pc25)
results["UMAP on top 25 PCs"] = (emb, t + pca25_time)

print("PCA PC1 + PC2 explain "
      f"{PCA(n_components=2).fit(X_std).explained_variance_ratio_.sum():.1%} of total variance")

# d. Scatter plots
superpops = {
    "African":     (["YRI", "LWK", "GWD", "MSL", "ESN", "ASW", "ACB"], "Oranges"),
    "European":    (["CEU", "TSI", "FIN", "GBR", "IBS"],               "Blues"),
    "East Asian":  (["CHB", "JPT", "CHS", "CDX", "KHV"],               "Greens"),
    "South Asian": (["GIH", "PJL", "BEB", "STU", "ITU"],               "Purples"),
    "American":    (["MXL", "PUR", "CLM", "PEL"],                      "Reds"),
}
colors = {}
for _, (codes, cmap) in superpops.items():
    shades = plt.get_cmap(cmap)(np.linspace(0.4, 0.95, len(codes)))
    colors.update(dict(zip(codes, shades)))
point_colors = np.array([colors[p] for p in pop])

fig, axes = plt.subplots(2, 3, figsize=(20, 13))
for ax, (name, (emb, secs)) in zip(axes.flat, results.items()):
    ax.scatter(emb[:, 0], emb[:, 1], c=point_colors, s=5, linewidths=0)
    ax.set_title(f"{name}  ({secs:.1f} s)", fontsize=13)
    ax.set_xticks([]); ax.set_yticks([])

legend_ax = axes.flat[-1]
legend_ax.axis("off")
handles = []
for group, (codes, _) in superpops.items():
    handles.append(Line2D([], [], ls="", label=group))
    handles += [Line2D([], [], marker="o", ls="", color=colors[c], label=f"  {c}") for c in codes]
legend_ax.legend(handles=handles, loc="center", ncol=2, fontsize=10, frameon=False,
                 title="Population (grouped by continent)", title_fontsize=11)
fig.suptitle("1000 Genomes subset: 2D projections coloured by population", fontsize=16)
plt.tight_layout()
plt.show()
```

e.

**Observations and best visual separation**

In the PCA plot, the populations mostly separate by continent, but there is noticeable overlap. The African populations separate along PC1 and the East Asian populations along PC2, while the European, South Asian, and American populations sit close together. The admixed populations (ACB, ASW, and the American groups) don’t form their own clusters. Instead, they fall between the main continental groups: ACB and ASW trail from the African cluster toward Europe, and the American samples are spread between Europe and South Asia. This makes sense because the first two components explain only about 12% of the total variance. Since PCA is a linear method, it captures the largest differences in the data, which here are the differences between continents. t-SNE separated the groups much more clearly than PCA, producing distinct clusters for the African, European, East Asian, and South Asian populations with little overlap. The admixed American populations, however, were scattered widely around the European cluster, with only PEL forming a small group of its own. The spacing and relative sizes of the clusters aren’t meaningful, because t-SNE focuses on preserving local neighborhoods rather than global structure. UMAP gave the best visual separation overall. The clusters were tight and well separated, and the American populations formed a continuous gradient from the European cluster toward PEL, which reflects their mixed ancestry. UMAP also preserved more of the overall layout than t-SNE. For example, Europe and the Americas stayed close together, and Africa and East Asia stayed far apart, which is consistent with the PCA plot. None of the methods clearly separated closely related populations within the same continent, such as CHB and JPT. With the top 25 PCs, there was some partial substructure: the East Asian cluster split into lighter (CHB, JPT, CHS) and darker (CDX, KHV) groups. We believe the limited within-continent resolution is mainly because our dataset contains only 868 variants, while the original paper used far more.

**Reducing to the top 25 PCs first**

Reducing the data to the top 25 principal components before applying t-SNE and UMAP made both methods faster. t-SNE dropped from about 41 to 35 seconds and UMAP from about 27 to 25 seconds, including the PCA step. It also produced cleaner plots, since PCA filters out much of the noise from variants that carry little population information. The clusters became tighter, and the American populations formed a clear gradient from Europe to PEL in both t-SNE and UMAP. The trade-off is that any variation outside the top 25 components is discarded, so smaller patterns in the data could be lost.

**Interpretability vs. performance trade-offs**

PCA is the fastest and most interpretable method, since its axes and distances have a clear meaning, but it provided the weakest separation. t-SNE and UMAP separated the populations much better, but they are slower, depend on hyperparameter choices, and can produce misleading cluster sizes and distances. UMAP offers a good balance because it preserves more global structure than t-SNE, and applying PCA first improves both speed and clarity. Overall, PCA is better suited for analyses where distances need to be interpreted, while t-SNE and UMAP are more useful for identifying and visualizing groups.
